```python
import os, sys

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from matplotlib import colormaps
from torchvision import transforms

sys.path.append('..')

import src.plotting as splot
import src.data_utils as sduti
```

```python
base = '../../NAIP-val_set/naip_restructured/'
images = os.listdir(base)
```

```python
which = 155
im = sduti.load_image( f'{base}{images[ which]}')
ima = sduti.to_numpy(im)
ima_ = ima / 255
```

```python
ax = splot.plot_single( ima)
_ = ax.set_title(ima_.var())
```

```python
r_m, g_m, b_m = ima_[:,:,0].mean(axis=0), ima_[:,:,1].mean(axis=0), ima_[:,:,2].mean(axis=0)
r_v, g_v, b_v = ima_[:,:,0].var(axis=0), ima_[:,:,1].var(axis=0), ima_[:,:,2].var(axis=0)

plt.bar(x=range(128),height= r_v, bottom= r_m - r_v/2, color='tab:red',alpha=0.3)
plt.bar(x=range(128),height= g_v, bottom= g_m - g_v/2, color='tab:green',alpha=0.3)
plt.bar(x=range(128),height= b_v, bottom= b_m - b_v/2, color='tab:blue',alpha=0.3)
#plt.ylim([0,1])
```
